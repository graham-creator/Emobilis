## **Chapter Two: Literature Review**

### 2.1 Introduction to the Chapter

This chapter reviews the existing literature on fraud detection in financial transactions, with a particular focus on machine learning applications in mobile money and similar digital payment systems.It establishes the theoretical foundation, summarises key empirical studies, and identifies the research gap that this project addresses.
### 2.2 Theoretical Background

Fraud detection is fundamentally a binary classification problem in supervised machine learning.The goal is to classify transactions as either legitimate or fraudulent.Due to the highly imbalanced nature of fraud datasets (fraud cases are typically less than 1% of total transactions), standard accuracy metrics are inadequate.Recall, Precision, and F1-score are commonly prioritised, with Recall being especially critical to minimise false negatives (missed fraud).Ensemble methods such as Random Forest and XGBoost have become popular because they handle class imbalance effectively and provide feature importance rankings.
### 2.3 Review of Related Work
Numerous studies have applied machine learning to fraud detection. Logistic Regression is frequently used as a baseline model due to its simplicity and interpretability (Sahin et al., 2013). Ensemble techniques have consistently shown superior performance. For example, Random Forest and XGBoost have been widely adopted in credit card and mobile money fraud detection tasks because they reduce overfitting and manage imbalanced data through techniques such as class weighting or SMOTE (Bhattacharyya et al., 2011; Adepoju et al., 2020). Recent research on synthetic mobile money datasets (such as PaySim) has shown that tree-based ensemble models outperform single classifiers in terms of Recall and F1-score (Lopez-Rojas et al., 2016). Studies also highlight the importance of features such as transaction amount, old and new balances, and transaction type for effective fraud prediction.
### 2.4 Data Sources and Datasets Used in Related Studies

Most fraud detection research uses either anonymised real credit card datasets or synthetic alternatives.The most widely used real dataset is the European credit card fraud dataset published by the Universite Libre de Bruxelles (ULB), which contains 284,807 transactions with 492 fraud cases.The PaySim dataset, developed by Lopez-Rojas et al. (2016) to simulate M-Pesa-style mobile money behaviour, has become the standard benchmark for mobile money fraud research because genuine transaction data is rarely available due to privacy and confidentiality restrictions.
### 2.5 Summary and Research Gap

The literature shows that classical machine learning models, particularly ensemble methods, are effective for fraud detection in imbalanced transactional data.However, most studies focus on credit card fraud or general synthetic data without specific emphasis on the performance comparison of Logistic Regression, Random Forest, and XGBoost on the PaySim dataset using Recall as the primary metric.This project addresses this gap by systematically comparing these three models on the PaySim dataset and providing clear feature importance insights.

## **Train/Test Split**

```python
# Splits 80% for training and 20% for testing, stratified to keep the same fraud ratio in both sets.
X = df.drop(columns=['isFraud'])
y = df['isFraud']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

print('Training rows:', len(X_train))
print('Test rows:', len(X_test))
print('Fraud rate in training set:', round(y_train.mean() * 100, 4), '%')
print('Fraud rate in test set:', round(y_test.mean() * 100, 4), '%')
```

### Markdown
- The dataset was split into 5,090,096 training rows and 1,272,524 test rows using an 
80:20 ratio. Stratified sampling was used to ensure both sets preserve the original 
fraud rate of 0.1291%. This is especially important with severely imbalanced data - 
without stratification a random split could place a disproportionate number of fraud 
cases in one set, making training or evaluation misleading.

## **Smote Technique**

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_train_sm, y_train_sm = smote.fit_resample(X_train, y_train)

print('Before SMOTE:', y_train.value_counts().to_dict())
print('After SMOTE:', y_train_sm.value_counts().to_dict())
```

### Markdown
- **SMOTE** (Synthetic Minority Oversampling Technique) was applied to the training set to 
address the severe class imbalance. Before SMOTE, the training set contained 5,083,526 
legitimate transactions and only 6,570 fraud cases. After SMOTE, both classes contain 
5,083,526 samples, giving the model an equal number of examples from each class to 
learn from.

- SMOTE works by generating synthetic fraud samples rather than simply duplicating 
existing ones. It selects a fraud case, finds its nearest neighbours in feature space, 
and creates new samples along the line between them. This produces more diverse 
synthetic examples compared to basic oversampling, which helps the model generalise 
better to unseen fraud patterns.

- Critically, SMOTE was applied only to the training set. The test set remains untouched 
with its original imbalanced distribution, so the evaluation metrics reflect real-world 
conditions where fraud genuinely represents 0.13% of transactions.

## **Feature Scaling**

```python
# Fits the scaler on the SMOTE training data only, then transforms both sets - no data leakage.
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_sm)
X_test_scaled = scaler.transform(X_test)

print('Scaling done.')
print('Training set shape:', X_train_scaled.shape)
print('Test set shape:', X_test_scaled.shape)
```

### Markdown
- StandardScaler was applied to normalise all numerical features to mean=0 and standard 
deviation=1. This is mandatory for Logistic Regression, which uses gradient-based 
optimisation that is sensitive to feature scale. Without scaling, features with large 
numeric ranges such as amount and account balances would dominate the model's 
coefficients simply because of their magnitude, not their predictive value.

- The scaler was fitted on the SMOTE training data only and then applied to the test set 
using transform. Fitting on the full dataset would constitute data leakage - the test 
set would have influenced the scaling parameters before evaluation, giving an 
artificially optimistic result.

- Random Forest and XGBoost do not strictly require scaling since they are tree-based 
models that split on individual feature thresholds. However, scaling was applied 
consistently across all three models to ensure a fair comparison.

## **4.3 Model Training Results**

## **Logistic Regression**

```python
# Trains Logistic Regression on the SMOTE balanced and scaled training data.
lr = LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42)
lr.fit(X_train_scaled, y_train_sm)
y_pred_lr = lr.predict(X_test_scaled)
```

```python
# Prints all four evaluation metrics.
print('Logistic Regression Results')
print('Recall:   ', round(recall_score(y_test, y_pred_lr), 4))
print('Precision:', round(precision_score(y_test, y_pred_lr), 4))
print('F1 Score: ', round(f1_score(y_test, y_pred_lr), 4))
print('Accuracy: ', round(accuracy_score(y_test, y_pred_lr), 4))
```

### Markdown
- Logistic Regression achieved a Recall of 0.9617, meaning it correctly identified 96.17% 
of all fraudulent transactions in the test set. This exceeds the minimum success 
threshold of 0.85 set in Section A and confirms that the model is effective at catching 
fraud cases.

- However, the Precision of 0.0249 is extremely low. This means that of all transactions 
the model flagged as fraud, only 2.49% were actually fraudulent. The model is generating 
a very high number of false alarms - legitimate transactions being incorrectly flagged 
as fraud. In a real deployment this would mean fraud analysts having to manually review 
a very large number of false positives, which is operationally costly.

- The F1 Score of 0.0486 reflects this imbalance between Recall and Precision. Accuracy 
of 0.9513 appears high but is misleading given the class imbalance - the model is 
approving most transactions as legitimate while casting a very wide net for fraud.

- Logistic Regression serves as the baseline for this project. Its high Recall but very 
low Precision is a known limitation of linear models on severely imbalanced data, even 
after SMOTE. The expectation is that ensemble models will improve Precision while 
maintaining strong Recall.

```python
# Shows the confusion matrix as a heatmap.
cm = confusion_matrix(y_test, y_pred_lr)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Legitimate', 'Fraud'])
disp.plot(cmap='Blues')
plt.title('Logistic Regression - Confusion Matrix')
plt.show()
```

### Markdown
- The confusion matrix confirms the pattern described above. Of the 1,643 actual fraud 
cases in the test set, the model correctly identified 1,580 (true positives) and missed 
only 63 (false negatives). This gives the high Recall of 0.9617.

- However, the model also flagged 61,860 legitimate transactions as fraud (false positives).

alarms. In a real mobile money system processing millions of transactions daily, this 
would generate an unmanageable volume of alerts for fraud analysts to review.

- The 1,210,601 true negatives confirm the model correctly clears the vast majority of 
legitimate transactions. The core challenge is reducing the 61,860 false positives 
without sacrificing the strong fraud detection rate, which is what the ensemble models 
in the next sections aim to address.

## **XGBoost**

```python
# Calculate scale_pos_weight to handle imbalance
scale = len(y_train_sm[y_train_sm==0]) / len(y_train_sm[y_train_sm==1])

xgb = XGBClassifier(n_estimators=100, scale_pos_weight=scale,
                     random_state=42, eval_metric='logloss',
                     n_jobs=-1)
xgb.fit(X_train_scaled, y_train_sm)
y_pred_xgb = xgb.predict(X_test_scaled)
```

### Markdown
- Trains XGBoost with scale_pos_weight set to the ratio of legitimate to fraud cases, giving the model a built-in mechanism to handle class imbalance.

```python
# Prints all four evaluation metrics.
print('XGBoost Results')
print('Recall:   ', round(recall_score(y_test, y_pred_xgb), 4))
print('Precision:', round(precision_score(y_test, y_pred_xgb), 4))
print('F1 Score: ', round(f1_score(y_test, y_pred_xgb), 4))
print('Accuracy: ', round(accuracy_score(y_test, y_pred_xgb), 4))
```

```python
# Shows the confusion matrix.
cm_xgb = confusion_matrix(y_test, y_pred_xgb)
disp = ConfusionMatrixDisplay(confusion_matrix=cm_xgb, display_labels=['Legitimate', 'Fraud'])
disp.plot(cmap='Blues')
plt.title('XGBoost - Confusion Matrix')
plt.show()
```

### Markdown
- XGBoost achieved the highest Recall of the three models at 0.9909, correctly identifying 
99.09% of all fraudulent transactions in the test set. Of the 1,643 actual fraud cases,

## **5.4 Limitations of the Study**
Several limitations affect the validity and generalisability of this project's findings. 
The most significant is that the PaySim dataset is synthetic. While it was designed to 
simulate real M-Pesa transaction behaviour, it may not capture the full complexity of 
fraud patterns in a live system, where fraudsters adapt their techniques in response 
to detection measures.

The analysis was conducted offline on a static snapshot of one month of simulated 
transactions. Real fraud detection systems operate in real time on streaming data, 
where model latency and computational efficiency become critical factors that were 
not evaluated here.

The project also did not explore more advanced sampling strategies beyond SMOTE, 
such as ADASYN or combination approaches using both oversampling and undersampling. 
It is possible that alternative sampling strategies could improve Precision without 
significantly reducing Recall.

Finally, hyperparameter tuning was limited due to the computational cost of running 
GridSearchCV on a dataset of over 6 million rows. A more exhaustive search may have 
produced better results, particularly for Random Forest and XGBoost.
## **5.5 Ethical Considerations**
Deploying a machine learning model for fraud detection carries significant ethical 
responsibilities. The most immediate concern is the impact of false positives on 
legitimate customers. When a genuine transaction is incorrectly flagged as fraud and 
blocked, the affected customer may be unable to pay bills, send money to family 
members, or access funds in an emergency. At scale, even a small false positive rate 
can affect thousands of people daily.

There is also a risk of systematic bias if the model's errors are not evenly distributed 
across customer segments. If certain transaction patterns associated with lower-income 
or rural users are more likely to be flagged as suspicious, the model could 
disproportionately disrupt the financial access of the most vulnerable customers - 
precisely the population that mobile money was designed to serve.

Transparency is another concern. Customers whose transactions are blocked have a right 
to understand why. Black-box models like Random Forest and XGBoost make it difficult 
to provide simple explanations for individual decisions. Any deployment should include 
a human review process for flagged transactions and a clear appeals mechanism for 
affected customers.

Finally, the model should not be treated as the final decision-maker. It should serve 
as a decision support tool that flags suspicious transactions for human review, rather 
than automatically blocking them without oversight.
